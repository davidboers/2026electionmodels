This is a useful cheat sheet to help translate between different units of swing and vote margin.

```python
import numpy as np
import pandas as pd

margin = [0, 1] + list(range(2, 100, 2))
swing = np.array(margin) / 2
table = pd.DataFrame(index=pd.Index(margin, name='Margin'))
table['Swing'] = swing
table['Winning Share'] = 50 + table['Swing']
table['Losing Share'] = 50 - table['Swing']
table
```
