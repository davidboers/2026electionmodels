```python
import matplotlib as plt
import geopandas

ss = geopandas.read_file('./gis/senate-2023-shape-file.zip')
sh = geopandas.read_file('./gis/house-2023-shape.zip')

counties = geopandas.read_file('./gis/GA-COUNTIES.json')
```

```python
dem_offense = [19, 25, 34, 40, 44, 45, 46, 47, 48, 49, 53, 81, 99, 103, 105, 120, 147, 151, 154, 164]
rep_offense = [50, 107, 108, 121, 128, 142, 145, 150, 168]

ax = sh.loc[sh['DATA'].isin(dem_offense)].plot(color='crimson', edgecolor='white')
sh.loc[sh['DATA'].isin(rep_offense)].plot(color='royalblue', ax=ax, edgecolor='white')
sh.loc[~(sh['DATA'].isin(dem_offense) | sh['DATA'].isin(rep_offense))].dissolve().plot(color='gray', ax=ax, autolim=False)
counties.plot(edgecolor='black', color='none', autolim=False, ax=ax)
```
